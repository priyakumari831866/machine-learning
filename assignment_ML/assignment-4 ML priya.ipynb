Q1. Employee Performance: Population vs. Sample Variance
A company records the performance scores of eight employees.

scores = [45, 50, 52, 55, 58, 60, 65, 75]
(a) Write Python code to calculate the mean, population variance, sample variance, population standard deviation, and sample standard deviation. Implement the variance and standard deviation formulas manually without using np.var() or np.std() for the calculations.
(b) Explain why the population and sample variance differ. Quantify the difference using the calculated results and recommend which variance formula should be used if these eight employees represent a sample of a much larger workforce.


```python
scores = [45, 50, 52, 55, 58, 60, 65, 75]

n = len(scores)
mean = sum(scores) / n

squared_deviations = [(x - mean) ** 2 for x in scores]
sum_squared_deviations = sum(squared_deviations)

population_variance = sum_squared_deviations / n

sample_variance = sum_squared_deviations / (n - 1)

population_sd = population_variance ** 0.5
sample_sd = sample_variance ** 0.5

print("Number of Employees:", n)
print("Mean:", mean)
print("Sum of Squared Deviations:", sum_squared_deviations)
print("Population Variance:", round(population_variance, 2))
print("Sample Variance:", round(sample_variance, 2))
print("Population Standard Deviation:", round(population_sd, 2))
print("Sample Standard Deviation:", round(sample_sd, 2))
```

Q2. Customer Spending: Comparing Data Spread
Use Seaborn's built-in tips dataset. Compare the total_bill and tip columns.

(a) Calculate the population variance and population standard deviation for both columns. Then calculate their means and coefficients of variation (CV). Use Python to compare their absolute and relative variability.

(b) Identify the five largest total_bill values and calculate how the mean, variance, and standard deviation change when those observations are temporarily excluded. Explain why variance can be more sensitive to extreme observations than the mean.

```python
import pandas as pd
import numpy as np
import seaborn as sns

tips = sns.load_dataset("tips")

def calculate_stats(series):
    data = series.dropna().tolist()
    n = len(data)

    mean = sum(data) / n

    variance = sum((x - mean) ** 2 for x in data) / n

    std_dev = np.sqrt(variance)

    cv = (std_dev / mean) * 100

    return mean, variance, std_dev, cv


for column in ["total_bill", "tip"]:
    mean, variance, std_dev, cv = calculate_stats(tips[column])

    print("\nColumn:", column)
    print("Mean:", round(mean, 3))
    print("Population Variance:", round(variance, 3))
    print("Population Standard Deviation:", round(std_dev, 3))
    print("Coefficient of Variation:", round(cv, 2), "%")
```

Q3. Build a Variance and Standard Deviation Calculator
Create a reusable Python program for the following datasets.

data_1 = [10, 20, 30, 40, 50]

data_2 = [20, 20, 20, 20, 20]
data_3 = [5, 10, 15, 20, 100]
data_4 = [12, 14, 16, 18, 20, 22]
(a) Implement separate functions to calculate the mean, population variance, sample variance, population standard deviation, and sample standard deviation without using NumPy's built-in statistical calculation functions. Correctly handle odd- and even-sized datasets.

(b) Test all four datasets and produce a formatted report. Identify the dataset with zero variability and the dataset most affected by an extreme observation. Justify your conclusions using the calculated variance and standard deviation.

```python
def calculate_mean(data):
    if len(data) == 0:
        raise ValueError("Data cannot be empty")
    return sum(data) / len(data)

def population_variance(data):
    mean = calculate_mean(data)
    total = sum((x - mean) ** 2 for x in data)
    return total / len(data)

def sample_variance(data):
    if len(data) < 2:
        raise ValueError("At least two values are required")
    mean = calculate_mean(data)
    total = sum((x - mean) ** 2 for x in data)
    return total / (len(data) - 1)

def population_std(data):
    return population_variance(data) ** 0.5


def sample_std(data):
    return sample_variance(data) ** 0.5

datasets = {
    "data_1": [10, 20, 30, 40, 50],
    "data_2": [20, 20, 20, 20, 20],
    "data_3": [5, 10, 15, 20, 100],
    "data_4": [12, 14, 16, 18, 20, 22]
}

print(f"{'Dataset':<10}{'Mean':>10}{'Pop Var':>12}"
      f"{'Sample Var':>13}{'Pop SD':>10}{'Sample SD':>12}")
print("-" * 67)

for name, data in datasets.items():
    print(
        f"{name:<10}"
        f"{calculate_mean(data):>10.2f}"
        f"{population_variance(data):>12.2f}"
        f"{sample_variance(data):>13.2f}"
        f"{population_std(data):>10.2f}"
        f"{sample_std(data):>12.2f}"
    )

data_3 = datasets["data_3"]
without_100 = [x for x in data_3 if x != 100]

print("\nData_3 before removing 100:")
print("Mean:", calculate_mean(data_3))
print("Population Variance:", population_variance(data_3))
print("Population SD:", round(population_std(data_3), 2))

print("\nData_3 after removing 100:")
print("Mean:", calculate_mean(without_100))
print("Population Variance:", population_variance(without_100))
print("Population SD:", round(population_std(without_100), 2))
```

Q4. Investigating the Effect of Mathematical Transformations
A researcher records these measurements:

measurements = [12, 15, 18, 20, 25, 30, 40]
(a) Create two new datasets: one by adding 10 to every observation and another by multiplying every observation by 4. Calculate the mean, population variance, and population standard deviation for all three datasets. Use Python to verify the mathematical relationships between the original and transformed statistics.

(b) Create a fourth dataset by multiplying every observation by \(-2\). Calculate its mean, variance, and standard deviation. Explain why a negative multiplier changes the direction of the values but does not make variance or standard deviation negative.



```python
from math import isclose

def calculate_mean(data):
    return sum(data) / len(data)

def calculate_variance(data):
    mean = calculate_mean(data)
    return sum((x - mean) ** 2 for x in data) / len(data)

def calculate_std(data):
    return calculate_variance(data) ** 0.5

measurements = [12, 15, 18, 20, 25, 30, 40]
add_10 = [x + 10 for x in measurements]
multiply_4 = [x * 4 for x in measurements]

multiply_minus_2 = [x * -2 for x in measurements]
datasets = {
    "Original": measurements,
    "Add 10": add_10,
    "Multiply by 4": multiply_4,
    "Multiply by -2": multiply_minus_2
}

print(f"{'Dataset':<16}{'Mean':>10}{'Variance':>12}{'Std Dev':>12}")
print("-" * 50)

for name, data in datasets.items():
    print(f"{name:<16}"
          f"{calculate_mean(data):>10.2f}"
          f"{calculate_variance(data):>12.2f}"
          f"{calculate_std(data):>12.2f}")

mean = calculate_mean(measurements)
variance = calculate_variance(measurements)
std = calculate_std(measurements)

assert isclose(calculate_mean(add_10), mean + 10)
assert isclose(calculate_variance(add_10), variance)
assert isclose(calculate_std(add_10), std)

assert isclose(calculate_mean(multiply_4), mean * 4)
assert isclose(calculate_variance(multiply_4), variance * 16)
assert isclose(calculate_std(multiply_4), std * 4)

assert isclose(calculate_mean(multiply_minus_2), mean * -2)
assert isclose(calculate_variance(multiply_minus_2), variance * 4)
assert isclose(calculate_std(multiply_minus_2), std * 2)

print("\nAll transformation checks passed!")
```

Q5. Outlier Detection and Robust Spread Analysis
A data analyst records daily sales, in thousands of rupees.

sales = [    12, 14, 15, 15, 16, 17, 18,    19, 20, 21, 22, 24, 25, 120]
(a) Write Python code to calculate the mean, median, population variance, and population standard deviation. Then identify potential outliers using the IQR method and report the quartiles, IQR, lower bound, upper bound, and flagged observations.



```python
import pandas as pd
sales = [12, 14, 15, 15, 16, 17, 18,
         19, 20, 21, 22, 24, 25, 120]

s = pd.Series(sales)

def calculate_mean(data):
    return sum(data) / len(data)

def calculate_median(data):
    data = sorted(data)
    n = len(data)

    if n % 2 == 1:
        return data[n // 2]
    return (data[n // 2 - 1] + data[n // 2]) / 2

def calculate_variance(data):
    mean = calculate_mean(data)
    return sum((x - mean) ** 2 for x in data) / len(data)

def calculate_std(data):
    return calculate_variance(data) ** 0.5

print("Mean:", round(calculate_mean(sales), 2))
print("Median:", calculate_median(sales))
print("Population Variance:", round(calculate_variance(sales), 2))
print("Population Standard Deviation:", round(calculate_std(sales), 2))

q1 = s.quantile(0.25)
q3 = s.quantile(0.75)
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outliers = [x for x in sales if x < lower or x > upper]

print("\nQ1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
print("Flagged Outliers:", outliers)
```

(b) Temporarily exclude the flagged outlier and recalculate the mean, median, variance, and standard deviation. Compare the percentage changes in variance and standard deviation, and recommend which statistics best describe typical daily sales while explaining why an extreme observation should be investigated before removal.

```python
clean_sales = [x for x in sales if x not in outliers]

old_mean = calculate_mean(sales)
old_median = calculate_median(sales)
old_variance = calculate_variance(sales)
old_std = calculate_std(sales)

new_mean = calculate_mean(clean_sales)
new_median = calculate_median(clean_sales)
new_variance = calculate_variance(clean_sales)
new_std = calculate_std(clean_sales)

variance_change = (new_variance - old_variance) / old_variance * 100
std_change = (new_std - old_std) / old_std * 100

print("Before Removing Outlier")
print("Mean:", round(old_mean, 2))
print("Median:", old_median)
print("Variance:", round(old_variance, 2))
print("Standard Deviation:", round(old_std, 2))

print("\nAfter Removing Outlier")
print("Mean:", round(new_mean, 2))
print("Median:", new_median)
print("Variance:", round(new_variance, 2))
print("Standard Deviation:", round(new_std, 2))

print("\nPercentage Changes")
print("Variance Change:", round(variance_change, 2), "%")
print("Standard Deviation Change:", round(std_change, 2), "%")
```
